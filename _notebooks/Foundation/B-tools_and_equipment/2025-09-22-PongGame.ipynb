Particle System - Copilot Carried

```python
function spawnParticles(x,y,dir){
  for(let i=0;i<8;i++){
    particles.push({
      x, y,
      vx: (Math.random()*2 + 0.5) * dir * 0.6,
      vy: (Math.random()-0.5)*3,
      size: Math.random()*3 + 1,
      life: 18 + Math.floor(Math.random()*8),
      maxLife: 18
    });
  }
}
```

The `spawnParticles` function generates short-lived particles at the collision point. Each particle object stores position, velocity, size, and remaining life; the render loop advances the particle positions and fades them out over time. This is a lightweight way to add polish without using external libraries.

Sound Effects

```python
function playSound(type){
  if(!audioCtx) audioCtx = new (window.AudioContext||window.webkitAudioContext)();
  const o = audioCtx.createOscillator();
  const g = audioCtx.createGain();
  o.connect(g); g.connect(audioCtx.destination);
  g.gain.value = 0.05;
  if(type==='hit'){ o.frequency.value = 880; }
  else if(type==='score'){ o.frequency.value = 220; }
  o.start();
  setTimeout(()=>{ o.stop(); }, 120);
}
```

The `playSound` helper creates short tones using the WebAudio API. We use an Oscillator and a Gain node to generate simple beeps: different frequencies and wave types (sine/sawtooth) correspond to hits versus scores. Note that many browsers require a user gesture before AudioContext will play sound; interacting with the page first (click or keypress) typically unlocks audio.

Pause Feature

```python
document.addEventListener('keydown', (e)=>{
  if(e.code === 'Space'){
    paused = !paused;
    e.preventDefault();
  }
});
```

This small listener toggles a `paused` flag when the Spacebar is pressed. The `update()` loop checks `paused` and early-returns so the game freezes while paused; this is a simple way to implement pause/resume without changing the main loop structure.
